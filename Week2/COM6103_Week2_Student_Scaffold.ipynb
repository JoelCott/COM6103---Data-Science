# 2. Conditional statements: making decisions

A conditional lets a program choose what to do.

```python
if condition:
    # runs when the condition is True
else:
    # runs when the condition is False
```

Python uses **indentation** to identify the block belonging to each condition.

## Real-world example: flagging an air-quality reading

Imagine a simplified sensor rule:

- below 35 → `"normal"`
- 35 to 54 → `"elevated"`
- 55 or above → `"high"`

````python
pm25 = 41

if pm25 < 35:
    status = "normal"
elif pm25 < 55:
    status = "elevated"
else:
    status = "high"

print("PM2.5:", pm25)
print("Status:", status)
````

### What happened, step by step

1. Python checks `pm25 < 35`.
2. If it is false, Python moves to the `elif`.
3. It checks `pm25 < 55`.
4. The first true branch runs.
5. Later branches are skipped.

This pattern is useful in data science when converting a numeric value into a category.

## Activity 5.2: `continue`

Data pipelines sometimes skip unusable records.

The list below contains missing values represented by `None`.

```python
values = [12.5, None, 15.0, 11.8, None, 14.2]
```

Use a loop and `continue` to skip `None`, then count and sum only the usable values.

````python
values = [12.5, None, 15.0, 11.8, None, 14.2]

total = 0
count = 0
i = 0

# TODO: loop through the list.
# Use continue when the current value is None.

while i < len(values):
    value = values[i]
    if value is None:
        i += 1
        continue
    total += value
    count += 1
    i += 1

print("Usable values:", count)
print("Total:", total)
if count > 0:
    print("Mean:", total / count)
````

# 9. Integrated mini-case: monitoring a small fleet of sensors

You are given a simplified list of hourly readings:

```python
readings = [72, 105, None, 64, 91, 118, 88, None, 77]
```

Rules:

- `None` means missing and should be skipped;
- 100 or above is `"high"`;
- 80 to 99 is `"elevated"`;
- below 80 is `"normal"`.

## Core task

Produce:

1. a list of labels for all non-missing readings;
2. the number of missing values;
3. the number of high readings;
4. the mean of all non-missing readings.

**Constraint:** use only the Python techniques covered so far. No pandas or NumPy.

````python
readings = [72, 105, None, 64, 91, 118, 88, None, 77]

labels = []
missing_count = 0
high_count = 0
total = 0
valid_count = 0

# TODO: process every reading.
# Step 1: detect and count None, then continue.
# Step 2: accumulate valid readings.
# Step 3: classify each valid reading.
# Step 4: count "high" readings.
for reading in readings:
    if reading is None:
        missing_count += 1
        labels.append("missing")
        continue
    total += reading
    valid_count += 1
    if reading >= 100:
        labels.append("high")
        high_count += 1
    else:
        labels.append("normal")

mean_reading = None
# TODO: calculate the mean safely
if valid_count > 0:
    mean_reading = total / valid_count

print("Labels:", labels)
print("Missing:", missing_count)
print("High:", high_count)
print("Mean:", mean_reading)
````

Validate automatically that:

- every number is between 1 and 59;
- no ticket contains duplicate numbers.

````python
# --// Lottery Ticket Generator \\--

# --// Dependencies
import random
import time
tickets = []

# --// Main Code
for i in range(5):
    ticket = random.sample(range(1, 51), 6)
    if ticket in tickets:
        print("Duplicate ticket generated, regenerating...")
        ticket = random.sample(range(1, 51), 6)
        
    tickets.append(ticket)
    print(tickets[i])

if tickets[0] == tickets[1]:
    print("Duplicate tickets found!", tickets[0], tickets[1])
else:
    print("\nAll tickets are unique.")

if tickets[0] == tickets[1] or tickets[0] == tickets[2] or tickets[0] == tickets[3] or tickets[0] == tickets[4]:
    print("Duplicate ticket found:", tickets[0])

winning_ticket = random.randint(0,4)
print("Winning ticket: loading...")

time.sleep(2)
print("Winning Ticket:", tickets[winning_ticket])
````

# 11. Final challenge: from rules to a tiny data-quality pipeline

A dataset arrives as a plain Python list:

```python
raw = [4, 7, -1, None, 12, 105, 9, 0, None, 18]
```

Build a new list called `clean` using these rules:

- skip `None`;
- skip negative values;
- stop processing completely if a value greater than 100 is encountered;
- keep all other values.

Then calculate the mean of the retained values.

This task deliberately combines `for`, `if`, `continue`, `break`, lists and accumulation.

````python
raw = [4, 7, -1, None, 12, 105, 9, 0, None, 18]
clean = []

# TODO: build clean using continue and break
for value in raw:
    if value is None:
        continue
    if value < 0 or value > 100:
        break
    clean.append(value)

mean = None
# TODO: calculate the mean of retained values
if clean:
    mean = sum(clean) / len(clean)

print("Clean:", clean)
print("Mean:", mean)
````
