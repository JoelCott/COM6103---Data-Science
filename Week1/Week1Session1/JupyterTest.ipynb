## 2. The data science lifecycle

A useful working lifecycle for this module is:

1. **Frame the problem** - define the scenario, analytical question and success criteria.
2. **Define data requirements** - decide what evidence is needed and what limitations matter.
3. **Acquire data** - obtain data from files, APIs, databases or other legitimate sources.
4. **Store and retrieve** - organise data so that it can be accessed reliably.
5. **Clean and transform** - deal with quality problems and prepare data for analysis.
6. **Explore** - summarise distributions, relationships, anomalies and missingness.
7. **Analyse or model** - apply appropriate statistical or computational techniques.
8. **Evaluate** - test assumptions, performance, uncertainty, bias and limitations.
9. **Communicate** - present evidence clearly through narrative, visualisation and professional outputs.
10. **Govern and reproduce** - document provenance, ethics, legal basis, code, versions and decisions.

This is **not a one-way pipeline**. Real investigations loop backwards when new evidence reveals a problem with the question, the data or the chosen method.

### Worked scenario: study-space occupancy

A university wants to understand whether four study spaces are being used effectively. The facilities team has a small set of aggregated observations containing room capacity, occupied seats, temperature and noise level.

A reasonable starting question is:

> **Which study spaces appear most heavily used in the observed sessions, and what further evidence would be needed before making a facilities decision?**

Notice that this question is deliberately cautious. A tiny observational dataset can support description, but it cannot by itself justify a causal claim or a major operational decision.

````python
import pandas as pd
import numpy as np

records = [
    {"session": "Mon 09:00", "space": "Library A", "capacity": 40, "occupied": 32, "temperature_c": 20.4, "noise_db": 42},
    {"session": "Mon 09:00", "space": "Library B", "capacity": 30, "occupied": 27, "temperature_c": 21.2, "noise_db": 48},
    {"session": "Mon 09:00", "space": "Study Hub", "capacity": 24, "occupied": 18, "temperature_c": 22.1, "noise_db": 51},
    {"session": "Mon 09:00", "space": "Atrium", "capacity": 50, "occupied": 29, "temperature_c": 19.8, "noise_db": 55},
    {"session": "Mon 14:00", "space": "Library A", "capacity": 40, "occupied": 36, "temperature_c": 21.0, "noise_db": 45},
    {"session": "Mon 14:00", "space": "Library B", "capacity": 30, "occupied": 29, "temperature_c": 21.7, "noise_db": 52},
    {"session": "Mon 14:00", "space": "Study Hub", "capacity": 24, "occupied": 22, "temperature_c": 22.5, "noise_db": np.nan},
    {"session": "Mon 14:00", "space": "Atrium", "capacity": 50, "occupied": 41, "temperature_c": 20.3, "noise_db": 61},
    {"session": "Tue 09:00", "space": "Library A", "capacity": 40, "occupied": 25, "temperature_c": 20.1, "noise_db": 40},
    {"session": "Tue 09:00", "space": "Library B", "capacity": 30, "occupied": 24, "temperature_c": 20.8, "noise_db": 46},
    {"session": "Tue 09:00", "space": "Study Hub", "capacity": 24, "occupied": 20, "temperature_c": 21.9, "noise_db": 50},
    {"session": "Tue 09:00", "space": "Atrium", "capacity": 50, "occupied": 35, "temperature_c": 19.9, "noise_db": 58},
]

df = pd.DataFrame(records)
df["occupancy_rate"] = df["occupied"] / df["capacity"]
df
````

The next two cells use pandas and Matplotlib. At Week 1 you do **not** need to memorise this syntax. Run the cells and focus on the evidence they produce.

````python
print("Rows, columns:", df.shape)
print("\nMissing values by column:")
display(df.isna().sum())

print("\nNumeric summary:")
display(df.select_dtypes(include="number").describe().round(2))
````

````python
import matplotlib.pyplot as plt

average_occupancy = (
    df.groupby("space", as_index=False)["occupancy_rate"]
      .mean()
      .sort_values("occupancy_rate", ascending=False)
)

display(average_occupancy)

ax = average_occupancy.plot(
    x="space",
    y="occupancy_rate",
    kind="bar",
    legend=False,
    rot=20,
    title="Average observed occupancy rate by study space"
)
ax.set_ylabel("Occupancy rate")
plt.tight_layout()
plt.show()
````

## 7. Reproducibility and portfolio setup

A reproducible project should make it possible for another person, including your future self, to understand where the data came from, what code was run, what assumptions were made and how outputs were produced.

### Suggested minimum portfolio structure

```text
COM6103-portfolio/
├── README.md
├── notebooks/
├── data/
│   ├── raw/
│   └── processed/
├── figures/
├── references/
└── .gitignore
```

This is a starting point, not a mandatory final architecture. The aim is to separate raw data, processed data, analysis notebooks, outputs and documentation clearly enough that another person can understand the project.

A useful `README.md` should eventually explain:

- what the repository contains;
- how to reproduce the work;
- where data came from;
- any installation requirements;
- ethical or legal constraints on the data;
- the status of the project; and
- who produced the work.

### Read-only Git diagnostic

Run the next cell. It checks for Git and an existing repository but does not initialise, commit or change anything.

````python
import subprocess
from pathlib import Path

def run_read_only(command):
    try:
        result = subprocess.run(
            command,
            capture_output=True,
            text=True,
            check=False
        )
        return (result.stdout or result.stderr).strip()
    except FileNotFoundError:
        return "Command not found."

print("Git:", run_read_only(["git", "--version"]))

current = Path.cwd().resolve()
repo_root = None
for candidate in [current, *current.parents]:
    if (candidate / ".git").exists():
        repo_root = candidate
        break

if repo_root is None:
    print("Repository check: this notebook is not currently inside a Git repository.")
else:
    print("Repository root:", repo_root)
    print(run_read_only(["git", "-C", str(repo_root), "status", "--short", "--branch"]))
````
